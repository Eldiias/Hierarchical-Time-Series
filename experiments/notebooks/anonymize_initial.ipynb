```python
import pandas as pd
```

```python
df = pd.read_csv('../../data/raw/plant_ids to be hidden.csv')
```

```python
df.drop(columns=['Unnamed: 0', 'Sold_to_name', 'Ship_to_name', 'Material_Availability_YearMonth', 'new_customer_id', 'Ordered_ratio_avg'], inplace=True)
```

```python
df.Plant = df.Plant.map({
    "TORRELAVEGA":'A',
    'DOMBASLE': 'B',
    'ROSIGNANO': 'C',
    'BERNBURG': 'D',
    'RHEINBERG': 'E',
    'DEVNYA': 'F',
}).fillna('NA')
```

```python
df.Packaging = df.Packaging.map({
    'Bulk': 'A',
    'Bags': 'B',
    'Big Bags': 'C',
    'Loose bags': 'D',
    'Pallets': 'E',
    'Sling bag': 'F'
})
```

```python
df.Calendar_YearMonth = pd.to_datetime(df.Calendar_YearMonth)
```

```python
df = df.rename(columns={'Calendar_YearMonth': 'Date'})
```

```python
zz = df[['Sold_to','Ship_to','Location','Product_Hier','Ship_to_Country','Packaging']].drop_duplicates().reset_index(drop=True).reset_index().rename(columns={'index': 'customer_id'})
```

```python
df2 = df.merge(zz, on=['Sold_to','Ship_to','Location','Product_Hier','Ship_to_Country','Packaging'], how='left', validate='m:1')
```

```python
df2.to_csv('../../data/raw/plant_ids.csv', index=False)
```
